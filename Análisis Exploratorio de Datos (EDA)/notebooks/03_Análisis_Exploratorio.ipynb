# **Estadística descriptiva**

```python
import pandas as pd
df = pd.read_csv( "/content/familias_renta_ciudadana_clean.csv" )
```

# **Familias por departamento**

```python
familias_departamento = ( df["nombredepartamentoatencion"] .value_counts() .head(10) )
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
familias_departamento.plot(kind='bar')
plt.title('Departamentos por Número de Familias')
plt.xlabel('Nombre del Departamento')
plt.ylabel('Número de Familias')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# **Familias por municipio**

```python
familias_municipio = ( df["nombremunicipioatencion"] .value_counts() .head(10) )
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 7))
familias_municipio.plot(kind='bar', color='skyblue')
plt.title('Municipios por Número de Familias')
plt.xlabel('Nombre del Municipio')
plt.ylabel('Número de Familias')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
